# Importação dos dados

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_1.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science/refs/heads/main/base-de-dados-challenge-1/loja_4.csv"

loja1 = pd.read_csv(url)
loja2 = pd.read_csv(url2)
loja3 = pd.read_csv(url3)
loja4 = pd.read_csv(url4)

loja1.head()
```

```python
loja1.head()
```

```python
## Unificar em um só DataFrame


loja1['Loja'] = 'Loja 1'
loja2['Loja'] = 'Loja 2'
loja3['Loja'] = 'Loja 3'
loja4['Loja'] = 'Loja 4'

df = pd.concat([loja1, loja2, loja3, loja4], ignore_index=True)
df
```

```python
print("Dados unificados carregados! Shape:", df.shape)
df.dtypes
```

#1. Análise do faturamento


```python
faturamento_lojas = df.groupby('Loja')['Preço'].sum().sort_values(ascending=False)
print ('\n=== FATURAMENTO TOTAL POR LOJA (soma de Preço) ===')
print (faturamento_lojas)
```

```python
faturamento_lojas = pd.DataFrame (faturamento_lojas)
faturamento_lojas = faturamento_lojas.assign (Percentual= faturamento_lojas['Preço'] / faturamento_lojas['Preço'].sum() * 100)
faturamento_lojas['Percentual'] = faturamento_lojas['Percentual'].map('{:.2f}%'.format)
faturamento_lojas
```

```python
# Agrupar por Loja e Categoria do Produto e somar o Preço
faturamento_categoria_loja = df.groupby(['Loja', 'Categoria do Produto'])['Preço'].sum()

# Obter o top 3 categorias por faturamento para cada loja
faturamento_categoria_loja_top3_series = faturamento_categoria_loja.groupby(level=0).nlargest(3)

# Calcular o faturamento total geral
faturamento_total_geral = df['Preço'].sum()

# Calcular a participação percentual de cada fatia (Loja x Top Categoria) no faturamento total geral
faturamento_percentual_total_top3 = (faturamento_categoria_loja_top3_series / faturamento_total_geral) * 100
```

#2. Vendas por Categoria

```python
# Vendas por Categoria por loja Número de Vendas

vendas_por_categoria_por_loja = df.groupby(['Loja', 'Categoria do Produto']).size()
print ('\n=== VENDAS POR CATEGORIA POR LOJA ===')
vendas_por_categoria_por_loja
```

```python
# Categoria mais popular em cada loja  (Top 3 por NÚMERO DE VENDAS por Loja)
print ('\n=== CATEGORIA MAIS VENDIDOS POR LOJA (Top 3, por contagem de transação) ===')
categoria_top = df.groupby(['Loja', 'Categoria do Produto']).size().groupby(level=0).nlargest(3)
df_categoria_top = pd.DataFrame(categoria_top)


df_categoria_top.rename(columns={0: 'Contagem de Transação'}, inplace=True)
df_categoria_top



```

```python
##Categorias mais vendidas (Top 3 por faturamento, geral)

categorias_top = df.groupby('Categoria do Produto')['Preço'].sum().nlargest(10)
print ('\n=== CATEGORIAS MAIS VENDIDAS (GERAL, por faturamento) ===')
categorias_top
```

#3. Média de Avaliação das Lojas

```python
#Avaliação média por loja
avaliacao_media = df.groupby('Loja')['Avaliação da compra'].mean().sort_values(ascending=True)
print ('\n=== AVALIAÇÃO MÉDIA POR LOJA ===')
avaliacao_media = round(avaliacao_media, 2)
avaliacao_media
```

```python
df_avaliacao_media = pd.DataFrame(avaliacao_media)
tabela_markdown = df_avaliacao_media.to_markdown(index=True)
print (tabela_markdown)
```

#4. Produtos Mais e Menos Vendidos

```python
# Produtos mais vendidos (Top 3 por NÚMERO DE VENDAS por Loja)
print ('\n=== PRODUTOS MAIS VENDIDOS POR LOJA (Top 3, por contagem de transação) ===')
produtos_top = df.groupby(['Loja', 'Produto']).size().groupby(level=0).nlargest(3)
print (produtos_top)
```

```python
# Produtos menos vendidos (Top 3 por NÚMERO DE VENDAS por Loja)
print ('\n=== PRODUTOS MENOS VENDIDOS POR LOJA (Top 3, por contagem de transação) ===')
produtos_bottom = df.groupby(['Loja', 'Produto']).size().groupby(level=0).nsmallest(3)
produtos_bottom
```

#5. Frete Médio por Loja

```python
# Frete médio por loja
frete_medio = df.groupby('Loja')['Frete'].mean().sort_values(ascending=True)
print ('\n=== FRETE MÉDIO POR LOJA ===')
frete_medio
```

#6. GRÁFICOS

```python
# GRAFICO 1 - Setores - Faturamento por Loja

plt.style.use('seaborn-v0_8-whitegrid')
faturamento_data = faturamento_lojas['Preço']

colors = plt.cm.Pastel1(range(len(faturamento_data)))


explode = [0.1 if loja == faturamento_data.idxmax() else 0 for loja in faturamento_data.index]

plt.figure(figsize=(10, 8))
plt.pie(faturamento_data, explode=explode, labels=faturamento_data.index, autopct='%1.1f%%',
        shadow=True, startangle=140, colors=colors, wedgeprops={'edgecolor': 'white'})

plt.title('Participação de cada Loja no Faturamento Total (%)', fontsize=20, ha='center')


plt.text(1.05, 0.7, "Faturamento por Loja:", horizontalalignment='left', verticalalignment='center', transform=plt.gca().transAxes, fontsize=12, fontweight='bold')
for i, (loja, faturamento) in enumerate(faturamento_data.items()):
    plt.text(1.05, 0.65 - i * 0.05, f"{loja}: R$ {faturamento:,.2f}", horizontalalignment='left', verticalalignment='center', transform=plt.gca().transAxes, fontsize=10)

plt.tight_layout(rect=[0, 0, 0.9, 1])
plt.savefig('faturamento_por_loja_pie.png')
plt.show()
```

```python
# Gráfico: Participação Percentual do Faturamento das Top 3 Categorias por Loja (no Faturamento Total Geral) - Empilhado


faturamento_percentual_total_top3_df = faturamento_percentual_total_top3.reset_index(level=0, drop=True)
faturamento_percentual_total_top3_df = faturamento_percentual_total_top3_df.reset_index()
faturamento_percentual_total_top3_df.columns = ['Loja', 'Categoria do Produto', 'Percentual Faturamento Total']
faturamento_percentual_total_top3_unstacked = faturamento_percentual_total_top3_df.pivot(
    index='Loja', columns='Categoria do Produto', values='Percentual Faturamento Total'
).fillna(0) # Preencher valores ausentes com 0

plt.figure(figsize=(15, 8))
ax = faturamento_percentual_total_top3_unstacked.plot(kind='bar', stacked=True)

plt.title('% de Faturamento Total pelas Top 3 Categorias (por Loja)')
plt.xlabel('Loja')
plt.ylabel('Participação Percentual no Faturamento Total Geral (%)')
plt.xticks(rotation=0)
plt.legend(title='Categoria do Produto', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('faturamento_percentual_total_top3_unstacked.png')
plt.show()
```

```python
# Gráfico 3: Dispersão - Avaliação vs. Faturamento por Loja (bolhas proporcionais ao faturamento.)

import pandas as pd
import matplotlib.pyplot as plt

# Combinar os dados de avaliação média e faturamento em um DataFrame para garantir alinhamento
# Selecionar apenas a coluna 'Preço' do DataFrame faturamento_lojas
df_combinado = pd.DataFrame({
    'Avaliação Média': avaliacao_media,
    'Faturamento': faturamento_lojas['Preço']
})


plt.figure(figsize=(10, 6))
scatter = plt.scatter(df_combinado['Avaliação Média'], df_combinado['Faturamento'] / 10000,
                      alpha=0.7, color=['green', 'blue', 'red', 'orange'])


for i, loja in enumerate(df_combinado.index):
    plt.annotate(loja, (df_combinado['Avaliação Média'].iloc[i], df_combinado['Faturamento'].iloc[i] / 10000),
                 xytext=(5, 5), textcoords='offset points')

plt.title('Avaliação Média vs. Faturamento por Loja')
plt.xlabel('Avaliação Média (1-5)')
plt.ylabel('Faturamento (escala para visual)')
plt.grid(True)
plt.savefig('avaliacao_faturamento_dispersao.png')
plt.show()
```

```python
# Grafico - Frete médio por loja (barras horizontais)
plt.figure(figsize=(12, 6))
ax = frete_medio.plot(kind='barh', color=['teal', 'salmon', 'lightblue', 'lightgreen'])

# Adicionar rótulos de dados às barras
for container in ax.containers:
    ax.bar_label(container, fmt='R$ %.2f', label_type='edge', padding=3)


plt.title('Frete Médio por Loja')
plt.xlabel('Frete Médio (R$)')
plt.ylabel('Loja')
plt.tight_layout()
plt.show()
plt.savefig('frete_medio.png')
```
